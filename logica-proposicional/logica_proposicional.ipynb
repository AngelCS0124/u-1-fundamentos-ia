```python
print(bool(0))
print(bool("hola"))
print(bool([]))
```

```python
def neg(p: bool) -> bool:
  return False if p else True

print(neg(True))
print(neg(False))
```

```python
def y(p: bool, q: bool) -> bool:
  return p and q

print(y(True, True))
print(y(True, False))
print(y(False, True))
print(y(False, False))
```

```python
def o(p: bool, q: bool) -> bool:
  return p or q

print(o(True, True))
print(o(True, False))
print(o(False, True))
print(o(False, False))
```

```python
def ox(p: bool, q: bool) -> bool:
  return neg(q) if p else q

print(ox(True, True))
print(ox(True, False))
print(ox(False, True))
print(ox(False, False))
```

```python
def impl(p: bool, q: bool) -> bool:
  return False if (p and not q) else True

print(impl(True, False))
print(impl(True, True))
print(impl(False, False))
```

```python
def ssi(p: bool, q: bool) -> bool:
  return p == q

print(ssi(True, True))
print(ssi(True, False))
print(ssi(False, False))
```

```python
print("=== AUDITORIA DE INGENIERIA DE PROMPTS ===\n")

llm_interpretacion_ambigua = lambda p, q, r: o(p, y(q, neg(r)))

llm_interpretacion_estructurada = lambda p, q, r: y(o(p, q), neg(r))

entorno_ataque = (True, False, True)

acceso_caso_a = llm_interpretacion_ambigua(*entorno_ataque)
acceso_caso_b = llm_interpretacion_estructurada(*entorno_ataque)

print("🚨 SIMULACIÓN DE ATAQUE DE SEGURIDAD 🚨")
print("Escenario: Admin = True | Contraseña Correcta = False | Cuenta Suspendida = True\n")
print(f"Resultado con Prompt Ambiguo : {'🔓 ACCESO CONCEDIDO (Fallo de seguridad)' \
                                        if acceso_caso_a else '🔒 ACCESO DENEGADO (Fallo de seguridad)'}}")
```
